```python
#This is to initialise the pandas and the dataset
from datasets import load_dataset
import pandas as pd

dataset = load_dataset("asafmak/personal-finance-ml-dataset", split="train")
df = dataset.to_pandas()

print("Dataset loaded successfully.")
print("Shape:", df.shape)
df.head()
```

3(a).

```python
#Here we find the number of columns and number of rows found in the dataset
rows, columns = df.shape
print(f"Number of rows: {rows}")
print(f"Number of columns: {columns}")
```

3(b)


```python
#Here we find the datatypes found in the dataset
print(df.dtypes)

print("\nDatatype counts:")
print(df.dtypes.value_counts())
```

3(c)

```python
#Here we convert to a pandas dataframe and check for null values
null_values = df.isna().sum()
display(null_values.to_frame("Null Values"))

if null_values.sum() == 0:
    print("Conclusion: The dataset is complete; there are no null values.")
else:
    print(f"Conclusion: The dataset is not complete; it contains {null_values.sum()} null values in total.")
```

3(d)

```python
#Here we combine the first 15 rows and last 20 rows and display that dataframe
df_sample = pd.concat(
    [df.head(15), df.tail(20)],
    ignore_index=True
)

print("Shape of df_sample:", df_sample.shape)
display(df_sample)
```
